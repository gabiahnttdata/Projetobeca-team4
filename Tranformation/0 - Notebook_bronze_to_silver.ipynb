```python
import pandas as pd
import os
import chardet
```

```python
# entrada
pasta = 'dataframe'  # '.' = diretorio atual, '..' => anterior
extensoes = ['csv'] # deixe em branco para todos

arquivos = [arq for arq in os.listdir(pasta) if arq.endswith(tuple(extensoes))]

# Cria um dicionário para armazenar os DataFrames
dataframes = {}

# Para cada arquivo na pasta
for i in arquivos:
    nome_variavel = os.path.splitext(i)[0]  # Nome do arquivo sem extensão
    dataframes[nome_variavel] = pd.read_csv(f'{pasta}/{i}', sep=';', encoding='latin1')
```
